$$

c) Determine the vector $p^{*}=\left(p_{1}^{*},\ldots,p_{n}^{*}\right)$ that results in maximum entropy. For this, one can use the Lagrange multiplier technique: subject to the constraint $p_{1}+\ldots+p_{n}=1$, the point $p^{*}$ that maximizes $H$ is the point at which $\nabla H(p)$ is orthogonal to the surface $\left\{p:p_{1}+\ldots+p_{n}=1\right\}$. \
